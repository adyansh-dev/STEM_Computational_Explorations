```python
# 1. Storying physics values in Python variables
u = 0       # initial velocity (m/s)
a = 1.6     # acceleration due to gravity on moon (m/s^2)
t = 15      # time (seconds)

# 2. Calculating the final velocity (v = u + at)
# In Python use '*' for multiplication
v = u + (a * t)

# 3. Calculating the distance fallen (s = ut + 0.5 * a * t^2)
# In Python use '**' for exponents
s = (u * t) + 0.5 * a * (t**2)

# 4. Telling Python to print the answers on my screen
print("--- PHYSICS SIMULATION RESULTS ---")
print("Final Velocity (v):", s, "m/s")
print("Distance Fallen (s):", s, "metres")
```
